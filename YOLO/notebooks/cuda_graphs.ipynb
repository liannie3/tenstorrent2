```python
#GRAPH CUDA EVENTS SHARE OF THE RUN
import csv
import matplotlib.pyplot as plot

#rows of most recent cuda events recorded in stage_timing.csv
rows = [r for r in csv.DictReader(open("../benchmarking/stage_timings.csv"))
        if r["pass"] == "substep_timing" and r["substep"] != "SUM_OF_STEPS"] #dont include sum of a substep stage
recent = max(r["timestamp"] for r in rows) #finx most recent
run = [r for r in rows if r["timestamp"]==recent] #only grabs substeps from most recent run

stages = {"pre_processing" : "#2a78d6", "inference" : "#eb6834", "post_processing" : "#1baf7a"} #dict to store each stage and their assoc. color
for stage, color in stages.items():
    share = [r for r in run if r["stage"]==stage]
    plot.barh([r["substep"] for r in share], [float(r["ms"]) for r in share], color = color, label = stage)
    
plot.xlabel("ms per image")
plot.title(f"Jetson Harness substep latency (batch {run[0]['batch_size']}, step {run[0]['step']})")
plot.show()
```
